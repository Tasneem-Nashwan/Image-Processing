```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread(r'E:\JERRY.jpg')
gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
enhanced=cv2.equalizeHist(gray)

plt.imshow(enhanced, cmap='gray')
plt.axis('off')
plt.show()
```

```python
hsv_img=cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
hsv_img=hsv_img.astype(np.float32)
hsv_img[:,:,1]=hsv_img[:,:,1]*1.5
hsv_img[:,:,2]=hsv_img[:,:,2]*1.4
hsv_img=np.clip(hsv_img, 0 ,255).astype(np.uint8)
bright_bgr=cv2.cvtColor(hsv_img, cv2.COLOR_HSV2BGR)
bright_rgb=cv2.cvtColor(bright_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(bright_rgb)
plt.axis('off')
plt.show()
```

```python
cropped_img =img[100:800, 200:1000]
cropped_rgb=cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB)

plt.imshow(cropped_rgb)
plt.title("Cropped Image")
plt.axis('off')
plt.show()
```

```python
resized_img=cv2.resize(img, (600,400))
cv2.imwrite('compressed_JERRY.jpg', resized_img,[cv2.IMWRITE_JPEG_QUALITY,50])
resized_rgb=cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB)

plt.imshow(resized_rgb)
plt.title("Resized Image")
plt.axis('off')
plt.show()
```
